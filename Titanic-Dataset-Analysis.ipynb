### Data Cleaning

```python
#Replace null values by mean in Age column
file["age"] = file["age"].fillna(file["age"].mean())
```

```python
file.select_dtypes(include = np.number).isnull().sum()
```

```python
#Find mode value in category columns
for col in file.select_dtypes(include='category').columns:
    print("Column Name:",col,  "-  Mode Value :",file[col].mode()[0])
```

```python
#Replace null values by mode in Deck column
file["deck"] = file["deck"].fillna(file["deck"].mode()[0])
```

```python

file.select_dtypes(include ='category').isnull().sum()
```

```python
# Find object mode value in dataset
for col in file.select_dtypes(include='object').columns:
    print("Column Name:", col ,  "-   Mode Value :",file[col].mode()[0])
```

```python
#Fill Mode values in object Columns
for col in file.select_dtypes(include='object').columns:
    file[col] = file[col].fillna(value=file[col].mode()[0])
```

```python
file.select_dtypes(include='object').isnull().sum()
```

```python
# Here is no any null values in dataset
file.isnull().sum()
```

```python
# Remove unnecessary column
file = file.drop(columns=['pclass'])

```

# Heatmap By Analysis



```python
plt.figure(figsize=(8,5))
sns.heatmap(file.isnull())
plt.show()
```

### **Survival Analysis by Class**

```python
summary = file.groupby('class')['survived'].agg(survived='sum',total='count').reset_index()
summary['not_survived'] = summary['total'] - summary['survived']
```

### **Summary Table**

```python
summary
```

## ***

###  Total Survived


```python
print("Total passengers:", len(file))
print("Total survived:", file["survived"].sum())
print("Total did not survive:", (file["survived"] == 0).sum())
```

```python
plt.figure(figsize=(9,5))
sns.countplot(x='survived',data=file, color='Blue')
plt.gca().set_facecolor('Black')
plt.xlabel("Survived" ,color='Black',fontweight='bold',fontsize=14)
plt.ylabel("Count" ,color='Black',fontweight='bold',fontsize=14)
plt.title("Total Survived", fontweight='bold')
plt.show()
```

# Survived by class

```python
chart_data = summary.melt(
    id_vars='class',
    value_vars =['survived' , 'not_survived'],
    var_name= 'Status',
    value_name='Count')

plt.figure(figsize=(9,8))
sns.barplot(data=chart_data,x='class',y='Count',hue='Status')

plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Survived vs Not Survived by Class")
plt.show()
```

### **Survived by Age**

```python
file["age_group"] = pd.cut(
    file["age"],
    bins=[0, 10, 18, 40, 60, 100],
    labels=["Child", "Teenager", "Adult", "Middle Age", "Senior"]
)

age_survival = file.groupby("age_group", observed=True)["survived"].mean().reset_index()
age_survival["survived"] = age_survival["survived"] * 100

plt.figure(figsize=(9,8))
sns.barplot(data=age_survival, x="age_group", y="survived")

plt.xlabel("Age Group",fontweight='bold',fontsize=14)
plt.ylabel("Survival Rate",fontweight='bold',fontsize=14)
plt.title("Survival Rate by Age Group",fontweight='bold',fontsize=20)
plt.show()
```

## **Survived by Gender**

```python
plt.figure(figsize=(9,7))
sns.countplot(x='sex',hue='survived',data=file)
plt.gca().set_facecolor('Black')
plt.xlabel("Gender" ,color='Black',fontweight='bold',fontsize=14)
plt.ylabel("Survived" ,color='Black',fontweight='bold',fontsize=14)
plt.title("Survived by Gender", fontweight='bold',fontsize=20)
plt.show()
```

## **Survived by Who**

```python
plt.figure(figsize=(9,5))
sns.barplot(x='who',y='survived',data=file)
plt.xlabel("WHO",color='Black',fontweight='bold',fontsize=16)
plt.ylabel("Survived",color='Black',fontweight='bold',fontsize=16)
plt.title("Survived by Who",fontweight='bold')
plt.show()
```

## **Alive by Embark_Town**


```python
plt.figure(figsize=(9,5))
sns.countplot(x="embark_town",hue="alive",data=file,alpha=0.7 ,palette=["red", "blue", "cyan"])
plt.gca().set_facecolor('BLack')
plt.title("Alive by Embark_Town",fontweight='bold',color='red')
plt.xlabel("Embark_Town", fontsize=14, fontweight='bold',color='red')
plt.ylabel("Class", fontsize=14, fontweight='bold', color='red')
plt.show()
```

## **Fare Statistics by Parch and Passenger Type**

```python
summary = file.groupby(["parch","who"], as_index=False)["fare"].agg(
    Count="count",
    Mean="mean",
    Sum="sum"
).reset_index()

summary 
```

## ***

```python
plt.figure(figsize=(9,5))
sns.barplot(data=summary, x='parch',y='Mean')
plt.xlabel("Parch",fontweight='bold',fontsize=15)
plt.ylabel("Average of Fare",fontweight='bold',fontsize=15)
plt.title("Parch vs Average Fare" ,fontweight='bold',fontsize=20)
plt.show()
```

## **Survival Summary by SibSp and Passenger Type**

```python
survival = file.groupby(["sibsp","who"])['survived'].agg(survived='sum',total='count').reset_index()
survival['not_survived'] = survival['total'] - survival['survived']
```

## Insights
* **Overall:** 342 of 891 passengers survived (**38.4%**).
* **Gender:** Females had a much higher survival rate than males.
* **Class:** **1st-class passengers** had the highest survival rate, While 3rd-class had the lowest.
* **Age:** The Children had relatively higher survival rates than older passengers.

* **Family Relationships:** SibSp and Parch were associated with survival. Passengers in **small family groups generally had higher survival rates**, while **large family groups had lower survival rates**.
Different family sizes showed different survival patterns.


* **Overall Conclusion:** The analysis shows that gender, passenger class, and age were important factors associated with survival on the Titanic. Female passengers, younger passengers, and passengers in higher classes generally had better survival outcomes.
